```python
import sys
import torch
from huggingface_hub import hf_hub_download
import matplotlib.pyplot as plt
import torchvision.utils as vutils
import pandas as pd
import torch.nn.functional as F

sys.path.insert(0, '..')

from src.models.generator import Generator
from src.models.mapper import Mapper
from src.training.loss import DirectCLIPloss
from src.utils import show_grid
```

```python
device = 'cuda' if torch.cuda.is_available() else 'cpu'

G = Generator('weights/ffhq.pkl', device=device)
loss_fn = DirectCLIPloss()
```

```python
styles = {
    'anime': 'mapper_an_anime_portrait.pt',
    'zombie': 'mapper_a_photo_of_a_person_as_a_zombie.pt',
    'vampire': 'mapper_a_realistic_vampire.pt'
}
mappers = {}

for style, filename in styles.items():

  mapper = Mapper(latent_dim=512, hidden_dim=1024, num_layers=4).to(device)

  ckpt = hf_hub_download(
    repo_id = 'tvictoria/styleclip-nada-ffhq',
    filename=filename
)

  checkpoint = torch.load(ckpt, map_location=device)

  mapper.load_state_dict(checkpoint["state_dict"])

  mapper.eval()

  mappers[style] = mapper
```

```python
seed = 42

torch.manual_seed(seed)

z_val = torch.randn(16, G.G.z_dim, device=device)

with torch.no_grad():

    w_val = G.mapping(z_val)

    img_before = G.synthesis(w_val)
```

```python
results = {}
metrics = {}
delta_norms = {}

source_text = 'a photo of a person'

with torch.no_grad():

  for style, mapper in mappers.items():

    delta = mapper(w_val)

    img_after = G.synthesis(w_val + delta)

    clip_loss = loss_fn(img_before, img_after, source_text, styles[style])

    l2_loss = delta.norm(dim=-1).mean()

    results[style] = img_after.cpu()

    metrics[style] = {
        'CLIP Loss': clip_loss.item(),
        'L2 Loss': l2_loss.item()
    }

    delta_norms[style] = delta.norm(dim=-1).cpu().numpy()
```

```python
metrics_df = pd.DataFrame(metrics).T
metrics_df
```

```python
before = F.interpolate(img_before.cpu(), size=256)
anime = F.interpolate(results['anime'], size=256)
zombie = F.interpolate(results['zombie'], size=256)
vampire = F.interpolate(results['vampire'], size=256)

fig, axes = plt.subplots(1, 4, figsize=(20,5))

images = [before, anime, zombie, vampire]
titles = ['Before', 'Anime', 'Zombie', 'Vampire']

for ax, imgs, title in zip(axes, images, titles):

    grid = vutils.make_grid((imgs.clamp(-1,1)+1)/2, nrow=4)

    ax.imshow(grid.permute(1,2,0))
    ax.set_title(title, fontsize=14)
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15,4))

for ax, style in zip(axes, delta_norms.keys()):

  ax.hist(delta_norms[style], bins=10)

  ax.set_title(style.capitalize())
  ax.set_xlabel(r"$||\Delta w||$")
  ax.set_ylabel("Count")

plt.tight_layout()
plt.show()
```
